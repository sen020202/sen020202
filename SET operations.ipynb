**SET operations**

```python

"""
Set Operations Program
This program demonstrates various set operations including intersection, union, and difference.
It allows users to input multiple sets and perform operations on them.
"""

def get_set_from_user(set_number):
    """
    Function to get a set from user input
    
    Args:
        set_number (int): The number/index of the set being input
    
    Returns:
        set: A set containing user-provided elements
    """
    print(f"\n--- Entering Set {set_number} ---")
    
    # Get the method of input from user
    print("Choose input method:")
    print("1. Enter elements separated by spaces")
    print("2. Enter elements one by one")
    
    while True:
        try:
            choice = int(input("Enter choice (1 or 2): "))
            if choice in [1, 2]:
                break
            else:
                print("Please enter 1 or 2")
        except ValueError:
            print("Please enter a valid number")
    
    user_set = set()  # Initialize empty set
    
    if choice == 1:
        # Method 1: Space-separated input
        while True:
            elements_input = input(f"Enter elements for Set {set_number} (separated by spaces): ").strip()
            if elements_input:  # Check if input is not empty
                # Split the input and convert to appropriate data types
                elements = elements_input.split()
                
                # Try to convert numbers to integers, keep strings as strings
                for element in elements:
                    try:
                        # Try to convert to integer first
                        user_set.add(int(element))
                    except ValueError:
                        try:
                            # If not integer, try float
                            user_set.add(float(element))
                        except ValueError:
                            # If neither, keep as string
                            user_set.add(element)
                break
            else:
                print("Please enter at least one element")
    
    else:
        # Method 2: One by one input
        print(f"Enter elements for Set {set_number} (press Enter without typing to finish):")
        while True:
            element = input("Enter element: ").strip()
            if element == "":  # Empty input means user is done
                if len(user_set) == 0:
                    print("Please enter at least one element")
                    continue
                break
            
            # Try to convert to appropriate data type
            try:
                user_set.add(int(element))
            except ValueError:
                try:
                    user_set.add(float(element))
                except ValueError:
                    user_set.add(element)
    
    return user_set

def display_set(set_data, set_name):
    """
    Function to display a set in a formatted way
    
    Args:
        set_data (set): The set to display
        set_name (str): Name/label for the set
    """
    print(f"{set_name}: {set_data}")

def find_intersection(sets_list):
    """
    Function to find intersection of multiple sets
    Intersection contains elements that are present in ALL sets
    
    Args:
        sets_list (list): List of sets
    
    Returns:
        set: Intersection of all sets
    """
    if not sets_list:  # Check if list is empty
        return set()
    
    # Start with the first set and find intersection with all others
    result = sets_list[0].copy()  # Make a copy to avoid modifying original
    
    # Iterate through remaining sets and find intersection
    for i in range(1, len(sets_list)):
        result = result.intersection(sets_list[i])
        # Alternative syntax: result = result & sets_list[i]
    
    return result

def find_union(sets_list):
    """
    Function to find union of multiple sets
    Union contains all unique elements from ALL sets
    
    Args:
        sets_list (list): List of sets
    
    Returns:
        set: Union of all sets
    """
    if not sets_list:  # Check if list is empty
        return set()
    
    # Start with empty set and add all elements from all sets
    result = set()
    
    # Iterate through all sets and add their elements
    for current_set in sets_list:
        result = result.union(current_set)
        # Alternative syntax: result = result | current_set
    
    return result

def find_difference(set1, set2):
    """
    Function to find difference between two sets
    Difference (set1 - set2) contains elements in set1 but not in set2
    
    Args:
        set1 (set): First set
        set2 (set): Second set
    
    Returns:
        set: Elements in set1 but not in set2
    """
    return set1.difference(set2)
    # Alternative syntax: return set1 - set2

def find_symmetric_difference(set1, set2):
    """
    Function to find symmetric difference between two sets
    Symmetric difference contains elements in either set but not in both
    
    Args:
        set1 (set): First set
        set2 (set): Second set
    
    Returns:
        set: Elements in either set but not in both
    """
    return set1.symmetric_difference(set2)
    # Alternative syntax: return set1 ^ set2

def display_menu():
    """
    Function to display the main menu options
    """
    print("\n" + "="*50)
    print("SET OPERATIONS MENU")
    print("="*50)
    print("1. Find Intersection of all sets")
    print("2. Find Union of all sets")
    print("3. Find Difference between sets")
    print("4. Find Symmetric Difference between two sets")
    print("5. Display all sets")
    print("6. Add more sets")
    print("7. Exit")
    print("="*50)

def main():
    """
    Main function that orchestrates the entire program
    """
    print("Welcome to Set Operations Program!")
    print("This program helps you perform various operations on sets.")
    
    # Get number of sets from user
    while True:
        try:
            num_sets = int(input("\nHow many sets do you want to work with? (minimum 2): "))
            if num_sets >= 2:
                break
            else:
                print("Please enter at least 2 sets to perform operations")
        except ValueError:
            print("Please enter a valid number")
    
    # Initialize list to store all sets
    sets_list = []
    
    # Get sets from user
    print(f"\nYou will now enter {num_sets} sets:")
    for i in range(num_sets):
        user_set = get_set_from_user(i + 1)
        sets_list.append(user_set)
        print(f"Set {i + 1} created: {user_set}")
    
    # Main program loop
    while True:
        display_menu()
        
        try:
            choice = int(input("Enter your choice (1-7): "))
        except ValueError:
            print("Please enter a valid number between 1 and 7")
            continue
        
        if choice == 1:
            # Find intersection of all sets
            print("\n--- INTERSECTION OPERATION ---")
            intersection_result = find_intersection(sets_list)
            print("Intersection of all sets (elements common to ALL sets):")
            print(f"Result: {intersection_result}")
            
            if len(intersection_result) == 0:
                print("No common elements found in all sets")
            else:
                print(f"Number of common elements: {len(intersection_result)}")
        
        elif choice == 2:
            # Find union of all sets
            print("\n--- UNION OPERATION ---")
            union_result = find_union(sets_list)
            print("Union of all sets (all unique elements from ALL sets):")
            print(f"Result: {union_result}")
            print(f"Total unique elements: {len(union_result)}")
        
        elif choice == 3:
            # Find difference between sets
            print("\n--- DIFFERENCE OPERATION ---")
            print("Available sets:")
            for i, s in enumerate(sets_list):
                print(f"Set {i + 1}: {s}")
            
            try:
                set1_idx = int(input("Enter first set number: ")) - 1
                set2_idx = int(input("Enter second set number: ")) - 1
                
                # Validate indices
                if 0 <= set1_idx < len(sets_list) and 0 <= set2_idx < len(sets_list):
                    diff_result = find_difference(sets_list[set1_idx], sets_list[set2_idx])
                    print(f"\nSet {set1_idx + 1} - Set {set2_idx + 1}:")
                    print(f"Elements in Set {set1_idx + 1} but not in Set {set2_idx + 1}:")
                    print(f"Result: {diff_result}")
                    
                    # Also show the reverse difference
                    reverse_diff = find_difference(sets_list[set2_idx], sets_list[set1_idx])
                    print(f"\nSet {set2_idx + 1} - Set {set1_idx + 1}:")
                    print(f"Elements in Set {set2_idx + 1} but not in Set {set1_idx + 1}:")
                    print(f"Result: {reverse_diff}")
                else:
                    print("Invalid set numbers!")
            except (ValueError, IndexError):
                print("Please enter valid set numbers")
        
        elif choice == 4:
            # Find symmetric difference between two sets
            print("\n--- SYMMETRIC DIFFERENCE OPERATION ---")
            print("Available sets:")
            for i, s in enumerate(sets_list):
                print(f"Set {i + 1}: {s}")
            
            try:
                set1_idx = int(input("Enter first set number: ")) - 1
                set2_idx = int(input("Enter second set number: ")) - 1
                
                # Validate indices
                if 0 <= set1_idx < len(sets_list) and 0 <= set2_idx < len(sets_list):
                    sym_diff_result = find_symmetric_difference(sets_list[set1_idx], sets_list[set2_idx])
                    print(f"\nSymmetric Difference between Set {set1_idx + 1} and Set {set2_idx + 1}:")
                    print("Elements in either set but not in both:")
                    print(f"Result: {sym_diff_result}")
                    print(f"Number of elements: {len(sym_diff_result)}")
                else:
                    print("Invalid set numbers!")
            except (ValueError, IndexError):
                print("Please enter valid set numbers")
        
        elif choice == 5:
            # Display all sets
            print("\n--- ALL SETS ---")
            for i, current_set in enumerate(sets_list):
                display_set(current_set, f"Set {i + 1}")
            print(f"\nTotal number of sets: {len(sets_list)}")
        
        elif choice == 6:
            # Add more sets
            print("\n--- ADD MORE SETS ---")
            try:
                additional_sets = int(input("How many additional sets do you want to add? "))
                if additional_sets > 0:
                    current_count = len(sets_list)
                    for i in range(additional_sets):
                        new_set = get_set_from_user(current_count + i + 1)
                        sets_list.append(new_set)
                        print(f"Set {current_count + i + 1} added: {new_set}")
                    print(f"\nTotal sets now: {len(sets_list)}")
                else:
                    print("Please enter a positive number")
            except ValueError:
                print("Please enter a valid number")
        
        elif choice == 7:
            # Exit program
            print("\nThank you for using Set Operations Program!")
            print("Goodbye!")
            break
        
        else:
            print("Invalid choice! Please enter a number between 1 and 7")
        
        # Pause before showing menu again
        input("\nPress Enter to continue...")

# Program execution starts here
if __name__ == "__main__":
    # This block ensures main() runs only when script is executed directly
    # Not when imported as a module
    main()
```
