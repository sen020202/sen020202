<span style = font-size:30px> __String Partitioning__ </span>

 The __partition()__ method in Python searches for a specified separator string within the target string and splits the string into three parts:

    The part before the separator,

    The separator itself,

    The part after the separator.

It always returns a tuple of three elements.

__Syntax__

string.partition(separator)

```python
str01 = "Natural language processing using Python and R and Java"
l= str01.partition('and')
print(l)
```

 __rpartition()__
 
Similar to partition() but splits at the last occurrence of the separator

    The part before the separator,

    The separator itself,

    The part after the separator.
    

```python
ll = str01.rpartition('and')
print(ll)
```

__split()__

Splits a string into a list of substrings.

You can specify how many splits to do using the maxsplit parameter.

Removes the separator from the result.

If the separator is not found → returns a list with the original string

```python
text = "one two three"
print(text.split(" "))  # ['one', 'two', 'three']

# where it found the space character it split the text there and create a separate enitty
```

```python
print(text.split(" ", 1))  # ['one', 'two three']

# maxsplit parameter set to 1, so it will stop splits after furst occurance of the partition
```

```python
print(text.split('o'))


""" The string "one two three" has 'o' in:

"one" → split before and after 'o'

"two" → split again at 'o'

So "one two three" splits like this:

'o' in "one" → split: ['', 'ne two three']

Next 'o' is in "two" → the part "ne two three" gets split into ['ne tw', ' three'] """

# since it splits before 'o' there is  an empty character at the begning of the result. 
```

__find() vs index() in Python__

Both find() and index() are used to locate a substring inside a string

— but they behave slightly differently when the substring is not found.

__find()__

      Returns the lowest index where the substring is found.

      Returns -1 if the substring is not found.

      Safe to use in condition checks.


__index()__
        Also returns the lowest index of the substring.

        But if the substring is not found, it raises a ValueError

Use find() when you're not sure the substring exists,

and 

index() when you're confident it does.

and 

want to handle errors if it doesn’t

__All / Any__

The all() method returns:

True - If all elements in a list are true

False - If any element in a list is false

The any() function returns True if any element in the list is True. If not, any() returns False


```python
l1 = [1,2,3,4,5,0]
```

```python
all(l1) # will return value false as one the value is false value=0.
```

```python
L2 = [1,2,3,4,True,False]
```

```python
all(L2) # Returns false as one value is false
```

```python
any(L2) # Returns True as many value 
```

```python
L3 = [1,2,3,True]
```

```python
all(L3) # Will return True as all items in the list are True
```
